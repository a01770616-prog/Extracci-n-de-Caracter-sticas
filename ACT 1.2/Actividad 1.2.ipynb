Existen muchas columnas con un solo o muy pocos valores nulos, decidimos no eliminarlos debido a que la base no cuenta con muchos datos.

```python
#Sustituir valores nulos por un string en  concreto 

info_forvia_p["Number"] =info_forvia_p["Number"].fillna("No description available")
info_forvia_p["Active"] =info_forvia_p["Active"].fillna("No description available")
info_forvia_p["Project Name"] =info_forvia_p["Project Name"].fillna("No Name")
info_forvia_p["Project Type"] =info_forvia_p["Project Type"].fillna("No description available")
info_forvia_p["Geographical scope"] =info_forvia_p["Geographical scope"].fillna("No description available")
info_forvia_p["Project manager"] =info_forvia_p["Project manager"].fillna("No description available")
info_forvia_p["State"] =info_forvia_p["State"].fillna("Not available")
info_forvia_p["Project size"] =info_forvia_p["Project size"].fillna("Not available")
info_forvia_p["Project organization"] =info_forvia_p["Project organization"].fillna("Not available")
info_forvia_p["Planned Go Live date"] =info_forvia_p["Planned Go Live date"].fillna("Not available")
info_forvia_p["BG"] =info_forvia_p["BG"].fillna("Not available")
info_forvia_p["Domain"] =info_forvia_p["Domain"].fillna("No description available")
info_forvia_p["Domain Path"] =info_forvia_p["Domain Path"].fillna("/")
info_forvia_p["Last WAR"] =info_forvia_p["Last WAR"].fillna("No description available")
info_forvia_p["Project Health"] =info_forvia_p["Project Health"].fillna("No color available")
info_forvia_p["Project type"] =info_forvia_p["Project type"].fillna("Regular")
info_forvia_p["Recurrent activity"] =info_forvia_p["Recurrent activity"].fillna("FALSO")
info_forvia_p["Actual start date"] =info_forvia_p["Actual start date"].fillna("NO DATE REGISTERED")
info_forvia_p["On-hold"] =info_forvia_p["On-hold"].fillna("No description available")
```

```python
#bathroom
info_forvia_p["Percent complete"]= info_forvia_p["Percent complete"].fillna(0) 
```

```python
#Identificar valores nulos por columna
valores_nulos=info_forvia_p.isnull().sum()
valores_nulos
```

```python
cuantitativas= info_forvia_p.select_dtypes(include=["float64"])
cualitativas = info_forvia_p.select_dtypes(include=["object"])
```

```python
#Realizamos diagrama de caja o bigote de cada columna del dataframe
fig = plt.figure(figsize =(15, 8))
cuantitativas.plot(kind='box', vert=False)
plt.title("Valores Atípicos del Dataframe")
plt.show() #dibujamos el diagrama
```

Usamos IQR ya que no sabemos bien cual sea la distribución al 100%, buscamos tener el mejor panorama por lo que esta opcion nos ayudara más. 

```python
#Método aplicando Cuartiles. Encuentro cuartiles 0.25 y 0.75
y=cuantitativas

percentile25=y.quantile(0.25) #Q1
percentile75=y.quantile(0.75) #Q3
iqr= percentile75 - percentile25

Limite_Superior_iqr= percentile75 + 1.5*iqr
Limite_Inferior_iqr= percentile25 - 1.5*iqr
print("Limite superior permitido", Limite_Superior_iqr)
print("Limite inferior permitido", Limite_Inferior_iqr)
```

```python
#Obtenemos datos limpios del Dataframe
data_iqr= cuantitativas[(y<=Limite_Superior_iqr)&(y>=Limite_Inferior_iqr)]
data_iqr
```

```python
#Corroboramos valores nulos del dataframe LIMPIO
valores_nulos=data_iqr.isnull().sum()
valores_nulos
```

Categorizar las siguientes variables: ” Percent complete” aplicando la regla de Sturges y desplegar en un gráfico las frecuencias de las categorías obtenidas.

**Percent complete**

```python
#Calculamos el numero total de la población "n"
info_forvia_p['Percent complete'].info()
n=245
```

```python
#Obtenemos el limite superior y el límite inferior de la columna objetivo
Max2=info_forvia_p['Percent complete'].max()
Min2=info_forvia_p['Percent complete'].min()
Limites2= [Min2, Max2]
Limites2
```

```python
#Calculamos el rango R
R2=Max2-Min2
R2
```

```python
#Calculamos el número de Intervalos de Clase "ni", aplicando la regla de Sturges
ni= 1+3.32*np.log10(n)
ni
```

```python
#Calculamos el Ancho del Intervalo "i"
i2=R2/ni
i2
```

```python
# Creamos 21 etiquetas para 22 intervalos generados por np.linspace
intervalos2 = np.linspace(-0.1, 100, 9)
categorias2 = [f"Categoría{i+1} ({round(intervalos2[i],2)} a {round(intervalos2[i+1],2)})" for i in range(len(intervalos2)-1)] 
categorias2
```

```python
#Finalmente creamos las categorías en la columna numérica
info_forvia_p['Percent complete']=pd.cut(x= info_forvia_p['Percent complete'], bins=intervalos2, labels= categorias2)
info_forvia_p['Percent complete']
```

```python
#Obtengo un análisis univariado de las variables categóricas
Tabla_freq2 = info_forvia_p['Percent complete'].value_counts().reset_index()
Tabla_freq2
```

```python
#Ajusto el indice de mi dataframe
Filtro_index12= Tabla_freq2.set_index('Percent complete')
Filtro_index12
```
